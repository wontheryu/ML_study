# 01. Linear Regression 실습
## 데이터 로드 및 EDA (탐색적 데이터 분석)
### EDA = Exploratory Data Analysis -> 데이터 분석 전에 데이터의 상태를 살피는 과정
🎯 EDA에서 반드시 해야 할 4가지 작업
1. 데이터의 형상 파악 (Structural Check)
데이터의 크기(Row, Column 개수)는 얼마인가?
변수들의 타입(수치형인가, 범주형인가)은 무엇인가?
결측치(Missing Value)가 존재하는가?
2. 기술 통계 확인 (Statistical Check)
평균, 중앙값, 최댓값, 최솟값 등을 확인합니다.
데이터가 한쪽으로 쏠려 있지는 않은지(Skewness), 이상치(Outlier)가 심하게 튀지 않는지 확인합니다.
3. 분포 및 관계 시각화 (Visualization)
히스토그램: 각 변수가 정규 분포를 따르는지 확인합니다 (선형 회귀의 정규성 가정과 연결).
산점도(Scatter Plot): 독립 변수와 종속 변수 사이에 정말 직선 형태의 관계(선형성)가 있는지 확인합니다.
히트맵(Heatmap): 변수들끼리 너무 친하지는 않은지(다중공선성 위험) 확인합니다.
4. 이상치(Outlier) 발견
평균적인 데이터 범위를 크게 벗어난 값이 있는지 확인합니다. 선형 회귀는 이상치에 매우 민감하여 직선의 방향을 왜곡시킬 수 있기 때문입니다.
